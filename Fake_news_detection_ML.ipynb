```python
# ============================================================
# INSTALL / IMPORT REQUIRED LIBRARIES
# ============================================================

import os
import re
import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import nltk

warnings.filterwarnings("ignore")


# ============================================================
#  DOWNLOAD NLTK RESOURCES
# ============================================================

try:
    STOP_WORDS = set(stopwords.words("english"))
except LookupError:
    nltk.download("stopwords")
    STOP_WORDS = set(stopwords.words("english"))

STEMMER = PorterStemmer()


# ============================================================
# CONFIGURATION
# ============================================================

TRAIN_PATH = "/content/train.csv"
TEST_PATH = "/content/test.csv"

OUTPUT_DIR = Path("/content/fake_news_results")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42


# ------------------------------------------------------------
# Label mapping
# ------------------------------------------------------------
# IMPORTANT:
# Verify this mapping against the official dataset documentation.
#
# Current assumed mapping:
#     0 = Fake
#     1 = Real
# ------------------------------------------------------------

LABEL_MAP = {
    0: "Fake",
    1: "Real"
}


# ------------------------------------------------------------
# TF-IDF configuration
# ------------------------------------------------------------

TFIDF_CONFIG = {
    "max_features": 50000,
    "ngram_range": (1, 2),
    "min_df": 2,
    "max_df": 0.95,
    "sublinear_tf": False
}


# ------------------------------------------------------------
# CountVectorizer configuration for RQ2
# ------------------------------------------------------------

COUNT_CONFIG = {
    "max_features": 50000,
    "ngram_range": (1, 2),
    "min_df": 2,
    "max_df": 0.95
}


# ------------------------------------------------------------
# Logistic Regression
# ------------------------------------------------------------

LR_CONFIG = {
    "C": 1.0,
    "max_iter": 2000,
    "solver": "liblinear",
    "random_state": RANDOM_STATE
}


# ------------------------------------------------------------
# Multinomial Naive Bayes
# ------------------------------------------------------------

NB_CONFIG = {
    "alpha": 1.0
}


# ------------------------------------------------------------
# Random Forest
# ------------------------------------------------------------

RF_CONFIG = {
    "n_estimators": 200,
    "random_state": RANDOM_STATE,
    "n_jobs": -1,
    "max_features": "sqrt"
}


# ------------------------------------------------------------
# RQ2 ablation
# ------------------------------------------------------------

RUN_RQ2_ABLATION = True


# ============================================================
# 4. HELPER FUNCTIONS
# ============================================================

def print_section(title):
    """Print a clearly formatted section heading."""
    print("\n")
    print("=" * 80)
    print(title)
    print("=" * 80)


def save_dataframe(df, filename):
    """Save dataframe to output directory."""
    path = OUTPUT_DIR / filename
    df.to_csv(path, index=False)
    print(f"Saved: {path}")
    return path


def save_json(data, filename):
    """Save dictionary as JSON."""
    path = OUTPUT_DIR / filename

    with open(path, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=4, default=str)

    print(f"Saved: {path}")
    return path


def safe_text(value):
    """Convert text safely to string."""
    if pd.isna(value):
        return ""

    return str(value)


# ============================================================
# LOAD DATA
# ============================================================

print_section("STEP 1 — LOADING DATA")

if not os.path.exists(TRAIN_PATH):
    raise FileNotFoundError(
        f"Training file not found: {TRAIN_PATH}"
    )

if not os.path.exists(TEST_PATH):
    raise FileNotFoundError(
        f"Testing file not found: {TEST_PATH}"
    )


train = pd.read_csv(TRAIN_PATH)
test = pd.read_csv(TEST_PATH)

print(f"Original training shape: {train.shape}")
print(f"Original testing shape : {test.shape}")


# ============================================================
#  CHECK REQUIRED COLUMNS
# ============================================================

print_section("STEP 2 — CHECKING DATASET STRUCTURE")

required_columns = {"text", "label"}

missing_train_columns = required_columns - set(train.columns)
missing_test_columns = required_columns - set(test.columns)

if missing_train_columns:
    raise ValueError(
        f"Training dataset is missing columns: "
        f"{missing_train_columns}"
    )

if missing_test_columns:
    raise ValueError(
        f"Testing dataset is missing columns: "
        f"{missing_test_columns}"
    )


# Keep only required columns
train = train[["text", "label"]].copy()
test = test[["text", "label"]].copy()

print("Required columns found.")
print("Columns:", list(train.columns))


# ============================================================
# INITIAL DATA AUDIT
# ============================================================

print_section("STEP 3 — INITIAL DATA AUDIT")

print("\nTraining information:")
print(train.info())

print("\nTesting information:")
print(test.info())


# ------------------------------------------------------------
# Missing values
# ------------------------------------------------------------

print("\nMissing values — Training:")
print(train.isnull().sum())

print("\nMissing values — Testing:")
print(test.isnull().sum())


# ------------------------------------------------------------
# Duplicate rows
# ------------------------------------------------------------

print("\nDuplicate complete rows:")
print(
    "Training:",
    train.duplicated().sum()
)

print(
    "Testing :",
    test.duplicated().sum()
)


# ------------------------------------------------------------
# Duplicate text values
# ------------------------------------------------------------

print("\nDuplicate text records:")

print(
    "Training:",
    train["text"].duplicated().sum()
)

print(
    "Testing :",
    test["text"].duplicated().sum()
)


# ------------------------------------------------------------
# Class distribution
# ------------------------------------------------------------

print("\nTraining class distribution:")
print(train["label"].value_counts(dropna=False))

print("\nTesting class distribution:")
print(test["label"].value_counts(dropna=False))


# ============================================================
#  REMOVE UNLABELLED RECORDS
# ============================================================

print_section("STEP 4 — REMOVING UNLABELLED RECORDS")

train_before = len(train)
test_before = len(test)

train = train.dropna(subset=["label"]).copy()
test = test.dropna(subset=["label"]).copy()

print(
    f"Training rows removed due to missing labels: "
    f"{train_before - len(train)}"
)

print(
    f"Testing rows removed due to missing labels: "
    f"{test_before - len(test)}"
)


# ============================================================
# VALIDATE LABELS
# ============================================================

print_section("STEP 5 — VALIDATING LABELS")

train["label"] = pd.to_numeric(train["label"], errors="coerce")
test["label"] = pd.to_numeric(test["label"], errors="coerce")

train = train.dropna(subset=["label"]).copy()
test = test.dropna(subset=["label"]).copy()

train["label"] = train["label"].astype(int)
test["label"] = test["label"].astype(int)

unique_train_labels = sorted(train["label"].unique())
unique_test_labels = sorted(test["label"].unique())

print("Training labels:", unique_train_labels)
print("Testing labels :", unique_test_labels)

if len(unique_train_labels) != 2:
    raise ValueError(
        "Training dataset does not contain exactly two classes."
    )

if len(unique_test_labels) != 2:
    raise ValueError(
        "Testing dataset does not contain exactly two classes."
    )

all_labels = sorted(set(unique_train_labels) | set(unique_test_labels))

if set(all_labels) != set(LABEL_MAP.keys()):
    print(
        "\nWARNING: LABEL_MAP does not exactly match "
        "the labels found in the data."
    )

print("\nCurrent label interpretation:")
for label, name in LABEL_MAP.items():
    print(f"{label} = {name}")


# ============================================================
#  REMOVE DUPLICATE TEXT WITHIN TRAINING DATA
# ============================================================

print_section("STEP 6 — REMOVING WITHIN-PARTITION DUPLICATES")

train_before_duplicates = len(train)
test_before_duplicates = len(test)

train = train.drop_duplicates(
    subset=["text"],
    keep="first"
).reset_index(drop=True)

test = test.drop_duplicates(
    subset=["text"],
    keep="first"
).reset_index(drop=True)

print(
    "Training duplicate texts removed:",
    train_before_duplicates - len(train)
)

print(
    "Testing duplicate texts removed:",
    test_before_duplicates - len(test)
)

print("Training shape:", train.shape)
print("Testing shape :", test.shape)


# ============================================================
# EXACT TRAIN/TEST OVERLAP CHECK
# ============================================================

print_section("STEP 7 — TRAIN/TEST OVERLAP CHECK")

train_texts = set(
    train["text"].astype(str)
)

test_texts = set(
    test["text"].astype(str)
)

overlap_texts = train_texts.intersection(test_texts)

print(
    f"Exact textual overlap detected: "
    f"{len(overlap_texts)}"
)


# ------------------------------------------------------------
# Save overlap records before removing them
# ------------------------------------------------------------

if len(overlap_texts) > 0:

    overlap_records = train[
        train["text"].astype(str).isin(overlap_texts)
    ].copy()

    save_dataframe(
        overlap_records,
        "01_train_test_overlap_removed.csv"
    )

else:

    overlap_records = pd.DataFrame(
        columns=train.columns
    )


# ------------------------------------------------------------
# Remove overlapping records FROM TRAIN ONLY
# ------------------------------------------------------------

train_before_overlap = len(train)

train = train[
    ~train["text"].astype(str).isin(overlap_texts)
].reset_index(drop=True)

print(
    "Training records removed because of exact "
    "train/test overlap:",
    train_before_overlap - len(train)
)

print(
    "Testing records retained:",
    len(test)
)

print("\nFinal raw dataset sizes:")
print("Training:", len(train))
print("Testing :", len(test))


# ============================================================
# SAVE DATA AUDIT SUMMARY
# ============================================================

audit_summary = pd.DataFrame({
    "Dataset": [
        "Original Training",
        "Original Testing",
        "Final Training",
        "Final Testing"
    ],
    "Rows": [
        train_before,
        test_before,
        len(train),
        len(test)
    ]
})

save_dataframe(
    audit_summary,
    "02_dataset_audit_summary.csv"
)


# ============================================================
# CLASS DISTRIBUTION AFTER CLEANING
# ============================================================

print_section("STEP 8 — FINAL CLASS DISTRIBUTION")

train_class_counts = train["label"].value_counts().sort_index()
test_class_counts = test["label"].value_counts().sort_index()

print("\nTraining class distribution:")
print(train_class_counts)

print("\nTesting class distribution:")
print(test_class_counts)


class_distribution = pd.DataFrame({
    "Train_Count": train_class_counts,
    "Test_Count": test_class_counts
}).fillna(0)

class_distribution["Train_Percentage"] = (
    class_distribution["Train_Count"]
    / class_distribution["Train_Count"].sum()
    * 100
)

class_distribution["Test_Percentage"] = (
    class_distribution["Test_Count"]
    / class_distribution["Test_Count"].sum()
    * 100
)

class_distribution["Class_Name"] = [
    LABEL_MAP.get(index, str(index))
    for index in class_distribution.index
]

save_dataframe(
    class_distribution.reset_index().rename(
        columns={"index": "label"}
    ),
    "03_class_distribution.csv"
)


# ============================================================
# EDA — TEXT LENGTH FEATURES
# ============================================================

print_section("STEP 9 — EXPLORATORY DATA ANALYSIS")

train_eda = train.copy()
test_eda = test.copy()


def word_count(text):
    return len(
        safe_text(text).split()
    )


def character_count(text):
    return len(
        safe_text(text)
    )


train_eda["word_count"] = (
    train_eda["text"]
    .apply(word_count)
)

test_eda["word_count"] = (
    test_eda["text"]
    .apply(word_count)
)

train_eda["character_count"] = (
    train_eda["text"]
    .apply(character_count)
)

test_eda["character_count"] = (
    test_eda["text"]
    .apply(character_count)
)


# ------------------------------------------------------------
# Descriptive statistics
# ------------------------------------------------------------

train_text_stats = (
    train_eda
    .groupby("label")
    [["word_count", "character_count"]]
    .agg(["count", "mean", "median", "std", "min", "max"])
)

print("\nTraining text statistics:")
print(train_text_stats)

train_text_stats.to_csv(
    OUTPUT_DIR / "04_training_text_statistics.csv"
)


# ============================================================
# EDA — CLASS DISTRIBUTION PLOT
# ============================================================

plt.figure(figsize=(8, 5))

class_names = [
    LABEL_MAP.get(x, str(x))
    for x in sorted(train["label"].unique())
]

sns.countplot(
    data=train,
    x="label"
)

plt.title("Training Dataset Class Distribution")
plt.xlabel("News Class")
plt.ylabel("Number of Articles")

plt.xticks(
    ticks=range(len(class_names)),
    labels=class_names
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "05_class_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
#  EDA — WORD COUNT DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

for label in sorted(train_eda["label"].unique()):

    subset = train_eda[
        train_eda["label"] == label
    ]

    plt.hist(
        subset["word_count"],
        bins=50,
        alpha=0.5,
        label=LABEL_MAP.get(label, str(label))
    )

plt.title("Article Word Count Distribution")
plt.xlabel("Number of Words")
plt.ylabel("Frequency")
plt.legend()

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "06_word_count_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
#  EDA — CHARACTER COUNT DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

for label in sorted(train_eda["label"].unique()):

    subset = train_eda[
        train_eda["label"] == label
    ]

    plt.hist(
        subset["character_count"],
        bins=50,
        alpha=0.5,
        label=LABEL_MAP.get(label, str(label))
    )

plt.title("Article Character Count Distribution")
plt.xlabel("Number of Characters")
plt.ylabel("Frequency")
plt.legend()

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "07_character_count_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# TEXT PREPROCESSING FUNCTION
# ============================================================

print_section("STEP 10 — TEXT PREPROCESSING")


def preprocess_text(text):
    """
    Text preprocessing pipeline:

    1. Convert to string
    2. Lowercase
    3. Remove non-alphabetic characters
    4. Tokenise
    5. Remove English stop words
    6. Keep alphabetic tokens
    7. Porter stemming
    8. Reconstruct cleaned text
    """

    text = safe_text(text)

    # Lowercase
    text = text.lower()

    # Keep alphabetic characters and whitespace
    text = re.sub(
        r"[^a-z\s]",
        " ",
        text
    )

    # Tokenisation
    tokens = text.split()

    # Stopword removal + alphabetic filtering
    tokens = [
        token
        for token in tokens
        if token.isalpha()
        and token not in STOP_WORDS
    ]

    # Porter stemming
    tokens = [
        STEMMER.stem(token)
        for token in tokens
    ]

    # Reconstruct cleaned document
    return " ".join(tokens)


# ============================================================
# APPLY PREPROCESSING
# ============================================================

print("Preprocessing training data...")

train["clean_text"] = (
    train["text"]
    .apply(preprocess_text)
)

print("Preprocessing testing data...")

test["clean_text"] = (
    test["text"]
    .apply(preprocess_text)
)


# ------------------------------------------------------------
# Remove empty documents after preprocessing
# ------------------------------------------------------------

empty_train = (
    train["clean_text"]
    .str.strip()
    .eq("")
)

empty_test = (
    test["clean_text"]
    .str.strip()
    .eq("")
)

print(
    "\nEmpty documents after preprocessing:"
)

print(
    "Training:",
    empty_train.sum()
)

print(
    "Testing :",
    empty_test.sum()
)

train = train[
    ~empty_train
].reset_index(drop=True)

test = test[
    ~empty_test
].reset_index(drop=True)


# ============================================================
# SAVE CLEANED DATA
# ============================================================

save_dataframe(
    train,
    "08_cleaned_training_data.csv"
)

save_dataframe(
    test,
    "09_cleaned_testing_data.csv"
)


# ============================================================
#  DISPLAY PREPROCESSING EXAMPLES
# ============================================================

print_section("PREPROCESSING EXAMPLES")

for i in range(min(5, len(train))):

    print("\nOriginal:")
    print(train.loc[i, "text"][:500])

    print("\nCleaned:")
    print(train.loc[i, "clean_text"][:500])

    print("-" * 80)


# ============================================================
# TF-IDF FEATURE EXTRACTION
# ============================================================

print_section("STEP 11 — TF-IDF FEATURE EXTRACTION")

print(
    "Fitting TF-IDF ONLY on training data..."
)

tfidf = TfidfVectorizer(
    **TFIDF_CONFIG
)

X_train_tfidf = tfidf.fit_transform(
    train["clean_text"]
)

X_test_tfidf = tfidf.transform(
    test["clean_text"]
)

y_train = train["label"]
y_test = test["label"]

print(
    "TF-IDF training matrix shape:",
    X_train_tfidf.shape
)

print(
    "TF-IDF testing matrix shape:",
    X_test_tfidf.shape
)

print(
    "Number of TF-IDF features:",
    len(tfidf.vocabulary_)
)


# ------------------------------------------------------------
# Save vocabulary information
# ------------------------------------------------------------

feature_names = tfidf.get_feature_names_out()

tfidf_features = pd.DataFrame({
    "feature_index": np.arange(
        len(feature_names)
    ),
    "feature": feature_names
})

save_dataframe(
    tfidf_features,
    "10_tfidf_feature_vocabulary.csv"
)


# ============================================================
#  MODEL DEFINITIONS
# ============================================================

models = {

    "Logistic Regression": LogisticRegression(
        **LR_CONFIG
    ),

    "Multinomial Naive Bayes": MultinomialNB(
        **NB_CONFIG
    ),

    "Random Forest": RandomForestClassifier(
        **RF_CONFIG
    )
}


# ============================================================
#  MODEL EVALUATION FUNCTION
# ============================================================

def evaluate_model(
    model,
    model_name,
    X_train,
    X_test,
    y_train,
    y_test,
    representation="TF-IDF"
):

    print("\n")
    print("=" * 80)
    print(f"TRAINING: {model_name}")
    print(f"REPRESENTATION: {representation}")
    print("=" * 80)

    # --------------------------------------------------------
    # Train
    # --------------------------------------------------------

    model.fit(
        X_train,
        y_train
    )

    # --------------------------------------------------------
    # Predictions
    # --------------------------------------------------------

    predictions = model.predict(
        X_test
    )

    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    weighted_precision = precision_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    weighted_recall = recall_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    weighted_f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    macro_precision = precision_score(
        y_test,
        predictions,
        average="macro",
        zero_division=0
    )

    macro_recall = recall_score(
        y_test,
        predictions,
        average="macro",
        zero_division=0
    )

    macro_f1 = f1_score(
        y_test,
        predictions,
        average="macro",
        zero_division=0
    )

    # --------------------------------------------------------
    # Confusion matrix
    # --------------------------------------------------------

    cm = confusion_matrix(
        y_test,
        predictions,
        labels=[0, 1]
    )

    tn, fp, fn, tp = cm.ravel()

    # --------------------------------------------------------
    # Classification report
    # --------------------------------------------------------

    report = classification_report(
        y_test,
        predictions,
        labels=[0, 1],
        target_names=[
            LABEL_MAP.get(0, "Class 0"),
            LABEL_MAP.get(1, "Class 1")
        ],
        zero_division=0
    )

    print("\nClassification Report:")
    print(report)

    print("\nConfusion Matrix:")
    print(cm)

    print("\nAccuracy:", round(accuracy, 4))
    print(
        "Weighted Precision:",
        round(weighted_precision, 4)
    )
    print(
        "Weighted Recall:",
        round(weighted_recall, 4)
    )
    print(
        "Weighted F1:",
        round(weighted_f1, 4)
    )

    print(
        "Macro Precision:",
        round(macro_precision, 4)
    )

    print(
        "Macro Recall:",
        round(macro_recall, 4)
    )

    print(
        "Macro F1:",
        round(macro_f1, 4)
    )

    # --------------------------------------------------------
    # Save classification report
    # --------------------------------------------------------

    safe_model_name = (
        model_name
        .lower()
        .replace(" ", "_")
    )

    safe_representation = (
        representation
        .lower()
        .replace(" ", "_")
    )

    report_path = (
        OUTPUT_DIR
        / f"report_{safe_representation}_{safe_model_name}.txt"
    )

    with open(
        report_path,
        "w",
        encoding="utf-8"
    ) as f:

        f.write(
            f"Model: {model_name}\n"
        )

        f.write(
            f"Representation: {representation}\n\n"
        )

        f.write(report)

        f.write("\n\nConfusion Matrix:\n")

        f.write(
            np.array2string(cm)
        )

    # --------------------------------------------------------
    # Confusion matrix plot
    # --------------------------------------------------------

    plt.figure(
        figsize=(7, 6)
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=[
            LABEL_MAP.get(0, "Class 0"),
            LABEL_MAP.get(1, "Class 1")
        ],
        yticklabels=[
            LABEL_MAP.get(0, "Class 0"),
            LABEL_MAP.get(1, "Class 1")
        ]
    )

    plt.title(
        f"{model_name} — Confusion Matrix"
    )

    plt.xlabel(
        "Predicted Label"
    )

    plt.ylabel(
        "Actual Label"
    )

    plt.tight_layout()

    cm_path = (
        OUTPUT_DIR
        / f"confusion_matrix_{safe_representation}_{safe_model_name}.png"
    )

    plt.savefig(
        cm_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    # --------------------------------------------------------
    # Save predictions
    # --------------------------------------------------------

    predictions_df = test[
        ["text", "label"]
    ].copy()

    predictions_df[
        "predicted_label"
    ] = predictions

    predictions_df[
        "actual_class"
    ] = predictions_df[
        "label"
    ].map(LABEL_MAP)

    predictions_df[
        "predicted_class"
    ] = predictions_df[
        "predicted_label"
    ].map(LABEL_MAP)

    predictions_df[
        "correct"
    ] = (
        predictions_df["label"]
        ==
        predictions_df["predicted_label"]
    )

    predictions_path = (
        OUTPUT_DIR
        / f"predictions_{safe_representation}_{safe_model_name}.csv"
    )

    predictions_df.to_csv(
        predictions_path,
        index=False
    )

    print(
        f"\nPredictions saved: {predictions_path}"
    )

    # --------------------------------------------------------
    # Return results
    # --------------------------------------------------------

    return {

        "Representation": representation,

        "Model": model_name,

        "Accuracy": accuracy,

        "Weighted Precision": weighted_precision,

        "Weighted Recall": weighted_recall,

        "Weighted F1": weighted_f1,

        "Macro Precision": macro_precision,

        "Macro Recall": macro_recall,

        "Macro F1": macro_f1,

        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp,

        "Test Samples": len(y_test),

        "Features": X_train.shape[1]
    }, predictions_df, cm, model


# ============================================================
# TRAIN ALL THREE MODELS USING TF-IDF
# ============================================================

print_section(
    "STEP 12 — TRAINING THREE CLASSICAL ML MODELS"
)

results = []

model_predictions = {}

trained_models = {}

confusion_matrices = {}


for model_name, model in models.items():

    result, prediction_df, cm, trained_model = (
        evaluate_model(
            model=model,
            model_name=model_name,
            X_train=X_train_tfidf,
            X_test=X_test_tfidf,
            y_train=y_train,
            y_test=y_test,
            representation="TF-IDF"
        )
    )

    results.append(result)

    model_predictions[model_name] = (
        prediction_df
    )

    trained_models[model_name] = (
        trained_model
    )

    confusion_matrices[model_name] = cm


# ============================================================
# MODEL COMPARISON
# ============================================================

print_section(
    "STEP 13 — MODEL PERFORMANCE COMPARISON"
)

results_df = pd.DataFrame(
    results
)

results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

print(
    results_df[
        [
            "Model",
            "Accuracy",
            "Weighted Precision",
            "Weighted Recall",
            "Weighted F1",
            "Macro Precision",
            "Macro Recall",
            "Macro F1"
        ]
    ]
)


# ------------------------------------------------------------
# Save raw metrics
# ------------------------------------------------------------

save_dataframe(
    results_df,
    "11_model_comparison_metrics.csv"
)


# ============================================================
#  PERCENTAGE VERSION FOR DISSERTATION
# ============================================================

percentage_results = results_df.copy()

metric_columns = [
    "Accuracy",
    "Weighted Precision",
    "Weighted Recall",
    "Weighted F1",
    "Macro Precision",
    "Macro Recall",
    "Macro F1"
]

for column in metric_columns:

    percentage_results[
        column
    ] = (
        percentage_results[column]
        * 100
    )

save_dataframe(
    percentage_results,
    "12_model_comparison_percentages.csv"
)


print("\nPerformance percentages:")

print(
    percentage_results[
        [
            "Model",
            "Accuracy",
            "Weighted Precision",
            "Weighted Recall",
            "Weighted F1",
            "Macro Precision",
            "Macro Recall",
            "Macro F1"
        ]
    ].round(2)
)


# ============================================================
# IDENTIFY BEST MODEL
# ============================================================

best_model_row = results_df.iloc[0]

best_model = best_model_row["Model"]

print("\n")
print("=" * 80)
print("BEST PERFORMING MODEL")
print("=" * 80)

print(
    "Model:",
    best_model
)

print(
    "Accuracy:",
    round(
        best_model_row["Accuracy"],
        4
    )
)

print(
    "Weighted F1:",
    round(
        best_model_row["Weighted F1"],
        4
    )
)


# ============================================================
# MODEL COMPARISON GRAPH
# ============================================================

print_section(
    "STEP 14 — MODEL COMPARISON GRAPH"
)

plot_metrics = [
    "Accuracy",
    "Weighted Precision",
    "Weighted Recall",
    "Weighted F1",
    "Macro F1"
]

plot_df = (
    results_df
    .set_index("Model")[plot_metrics]
)

plot_df.plot(
    kind="bar",
    figsize=(12, 7)
)

plt.title(
    "Comparison of Classical Machine Learning Models"
)

plt.xlabel(
    "Machine Learning Model"
)

plt.ylabel(
    "Score"
)

plt.ylim(
    0,
    1
)

plt.xticks(
    rotation=0
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR
    / "13_model_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
#  ACCURACY-ONLY GRAPH
# ============================================================

plt.figure(
    figsize=(9, 6)
)

plt.bar(
    results_df["Model"],
    results_df["Accuracy"]
)

plt.title(
    "Accuracy Comparison of Machine Learning Models"
)

plt.xlabel(
    "Model"
)

plt.ylabel(
    "Accuracy"
)

plt.ylim(
    0,
    1
)

plt.xticks(
    rotation=15
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR
    / "14_accuracy_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# F1 SCORE GRAPH
# ============================================================

plt.figure(
    figsize=(9, 6)
)

plt.bar(
    results_df["Model"],
    results_df["Weighted F1"]
)

plt.title(
    "Weighted F1-Score Comparison"
)

plt.xlabel(
    "Model"
)

plt.ylabel(
    "Weighted F1-Score"
)

plt.ylim(
    0,
    1
)

plt.xticks(
    rotation=15
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR
    / "15_f1_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# RQ1 ANALYSIS
# ============================================================

print_section(
    "RQ1 — EFFECTIVENESS OF CLASSICAL SUPERVISED ML"
)

print(
    """
RQ1:
To what extent are classical supervised ML algorithms helpful
in identifying fake news based on textual news data?
"""
)

print("\nModel performance:")

for _, row in results_df.iterrows():

    print(
        f"{row['Model']}: "
        f"Accuracy={row['Accuracy']:.4f}, "
        f"Weighted F1={row['Weighted F1']:.4f}"
    )

print(
    f"""
RQ1 interpretation:

All three classical supervised machine-learning algorithms were
trained using the same cleaned textual data and the same TF-IDF
feature representation. Their performance can therefore be compared
under a common experimental setup.

The highest-performing algorithm under the primary evaluation
criterion of accuracy was:

{best_model}

with an accuracy of {best_model_row['Accuracy']:.4f}.

The results provide empirical evidence regarding the usefulness of
classical supervised machine learning for binary fake-news
classification within this dataset.
"""
)


# ============================================================
# RQ3 ANALYSIS
# ============================================================

print_section(
    "RQ3 — MOST EFFECTIVE CLASSIFIER"
)

print(
    """
RQ3:
Which of Logistic Regression, Multinomial Naive Bayes and Random
Forest is most effective under traditional evaluation criteria?
"""
)

print(
    f"""
Based on the independent test-set results, the highest-performing
classifier according to accuracy was:

{best_model}

Accuracy:
{best_model_row['Accuracy']:.4f}

Weighted Precision:
{best_model_row['Weighted Precision']:.4f}

Weighted Recall:
{best_model_row['Weighted Recall']:.4f}

Weighted F1:
{best_model_row['Weighted F1']:.4f}

Macro F1:
{best_model_row['Macro F1']:.4f}
"""
)


# ============================================================
# FAKE NEWS-SPECIFIC ERROR ANALYSIS
# ============================================================

print_section(
    "STEP 15 — FAKE NEWS ERROR ANALYSIS"
)

error_analysis = []


for model_name, prediction_df in model_predictions.items():

    y_true = prediction_df["label"].values
    y_pred = prediction_df["predicted_label"].values

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    tn, fp, fn, tp = cm.ravel()

    # --------------------------------------------------------
    # Assuming:
    # 0 = Fake
    # 1 = Real
    #
    # False negative:
    # Actual Fake -> predicted Real
    # --------------------------------------------------------

    fake_recall = recall_score(
        y_true,
        y_pred,
        pos_label=0,
        zero_division=0
    )

    fake_precision = precision_score(
        y_true,
        y_pred,
        pos_label=0,
        zero_division=0
    )

    fake_missed_as_real = fn

    real_misclassified_as_fake = fp

    error_analysis.append({

        "Model": model_name,

        "Fake Precision": fake_precision,

        "Fake Recall": fake_recall,

        "Actual Fake Predicted Real": fake_missed_as_real,

        "Actual Real Predicted Fake": real_misclassified_as_fake,

        "Total Test Errors":
            (y_true != y_pred).sum()
    })


error_analysis_df = pd.DataFrame(
    error_analysis
)

print(
    error_analysis_df
)

save_dataframe(
    error_analysis_df,
    "16_fake_news_error_analysis.csv"
)


# ============================================================
# SAVE FALSE NEGATIVE ARTICLES
# ============================================================

print_section(
    "STEP 16 — SAVING MISCLASSIFIED ARTICLES"
)

for model_name, prediction_df in model_predictions.items():

    safe_model_name = (
        model_name
        .lower()
        .replace(" ", "_")
    )

    # Actual fake, predicted real
    fake_as_real = prediction_df[
        (prediction_df["label"] == 0)
        &
        (prediction_df["predicted_label"] == 1)
    ].copy()

    # Actual real, predicted fake
    real_as_fake = prediction_df[
        (prediction_df["label"] == 1)
        &
        (prediction_df["predicted_label"] == 0)
    ].copy()

    fake_as_real.to_csv(
        OUTPUT_DIR
        / f"fake_misclassified_as_real_{safe_model_name}.csv",
        index=False
    )

    real_as_fake.to_csv(
        OUTPUT_DIR
        / f"real_misclassified_as_fake_{safe_model_name}.csv",
        index=False
    )

    print(
        f"\n{model_name}:"
    )

    print(
        "Fake predicted as Real:",
        len(fake_as_real)
    )

    print(
        "Real predicted as Fake:",
        len(real_as_fake)
    )


# ============================================================
# RQ2 — TF-IDF VS COUNT VECTORIZER ABLATION
# ============================================================

if RUN_RQ2_ABLATION:

    print_section(
        "STEP 17 — RQ2 TF-IDF VS COUNTVECTORIZER ABLATION"
    )

    print(
        """
RQ2:
How does a uniform TF-IDF feature extraction model affect
fake-news detection success?
"""
    )

    # --------------------------------------------------------
    # CountVectorizer
    # --------------------------------------------------------

    print(
        "\nFitting CountVectorizer ONLY on training data..."
    )

    count_vectorizer = CountVectorizer(
        **COUNT_CONFIG
    )

    X_train_count = (
        count_vectorizer.fit_transform(
            train["clean_text"]
        )
    )

    X_test_count = (
        count_vectorizer.transform(
            test["clean_text"]
        )
    )

    print(
        "CountVectorizer training matrix:",
        X_train_count.shape
    )

    print(
        "CountVectorizer testing matrix:",
        X_test_count.shape
    )

    # --------------------------------------------------------
    # Run all three models with CountVectorizer
    # --------------------------------------------------------

    ablation_results = []

    # First, add the already obtained TF-IDF results
    for result in results:

        ablation_results.append({
            "Representation":
                "TF-IDF",

            "Model":
                result["Model"],

            "Accuracy":
                result["Accuracy"],

            "Weighted Precision":
                result["Weighted Precision"],

            "Weighted Recall":
                result["Weighted Recall"],

            "Weighted F1":
                result["Weighted F1"],

            "Macro Precision":
                result["Macro Precision"],

            "Macro Recall":
                result["Macro Recall"],

            "Macro F1":
                result["Macro F1"]
        })


    # --------------------------------------------------------
    # CountVectorizer models
    # --------------------------------------------------------

    count_models = {

        "Logistic Regression": LogisticRegression(
            **LR_CONFIG
        ),

        "Multinomial Naive Bayes": MultinomialNB(
            **NB_CONFIG
        ),

        "Random Forest": RandomForestClassifier(
            **RF_CONFIG
        )
    }


    for model_name, model in count_models.items():

        result, _, _, _ = evaluate_model(
            model=model,
            model_name=model_name,
            X_train=X_train_count,
            X_test=X_test_count,
            y_train=y_train,
            y_test=y_test,
            representation="CountVectorizer"
        )

        ablation_results.append({

            "Representation":
                "CountVectorizer",

            "Model":
                model_name,

            "Accuracy":
                result["Accuracy"],

            "Weighted Precision":
                result["Weighted Precision"],

            "Weighted Recall":
                result["Weighted Recall"],

            "Weighted F1":
                result["Weighted F1"],

            "Macro Precision":
                result["Macro Precision"],

            "Macro Recall":
                result["Macro Recall"],

            "Macro F1":
                result["Macro F1"]
        })


    # --------------------------------------------------------
    # Create ablation dataframe
    # --------------------------------------------------------

    ablation_df = pd.DataFrame(
        ablation_results
    )

    print(
        "\nTF-IDF vs CountVectorizer:"
    )

    print(
        ablation_df.round(4)
    )

    save_dataframe(
        ablation_df,
        "17_rq2_tfidf_ablation.csv"
    )


    # ========================================================
    #  RQ2 COMPARISON PLOT
    # ========================================================

    plt.figure(
        figsize=(12, 7)
    )

    for representation in [
        "TF-IDF",
        "CountVectorizer"
    ]:

        subset = ablation_df[
            ablation_df["Representation"]
            == representation
        ]

        plt.plot(
            subset["Model"],
            subset["Accuracy"],
            marker="o",
            label=representation
        )

    plt.title(
        "RQ2: TF-IDF vs CountVectorizer Accuracy"
    )

    plt.xlabel(
        "Machine Learning Model"
    )

    plt.ylabel(
        "Accuracy"
    )

    plt.ylim(
        0,
        1
    )

    plt.legend()

    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR
        / "18_rq2_tfidf_vs_count_accuracy.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


    # ========================================================
    #  RQ2 F1 COMPARISON
    # ========================================================

    plt.figure(
        figsize=(12, 7)
    )

    x = np.arange(
        len(count_models)
    )

    width = 0.35

    tfidf_values = []

    count_values = []

    model_names = list(
        count_models.keys()
    )

    for model_name in model_names:

        tfidf_value = ablation_df[
            (
                ablation_df["Representation"]
                == "TF-IDF"
            )
            &
            (
                ablation_df["Model"]
                == model_name
            )
        ]["Weighted F1"].iloc[0]

        count_value = ablation_df[
            (
                ablation_df["Representation"]
                == "CountVectorizer"
            )
            &
            (
                ablation_df["Model"]
                == model_name
            )
        ]["Weighted F1"].iloc[0]

        tfidf_values.append(
            tfidf_value
        )

        count_values.append(
            count_value
        )

    plt.bar(
        x - width / 2,
        tfidf_values,
        width,
        label="TF-IDF"
    )

    plt.bar(
        x + width / 2,
        count_values,
        width,
        label="CountVectorizer"
    )

    plt.xticks(
        x,
        model_names,
        rotation=15
    )

    plt.ylabel(
        "Weighted F1"
    )

    plt.xlabel(
        "Machine Learning Model"
    )

    plt.title(
        "RQ2: TF-IDF vs CountVectorizer Weighted F1"
    )

    plt.ylim(
        0,
        1
    )

    plt.legend()

    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR
        / "19_rq2_tfidf_vs_count_f1.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


    # ========================================================
    #  CALCULATE TF-IDF IMPROVEMENT
    # ========================================================

    rq2_rows = []

    for model_name in model_names:

        tfidf_row = ablation_df[
            (
                ablation_df["Representation"]
                == "TF-IDF"
            )
            &
            (
                ablation_df["Model"]
                == model_name
            )
        ].iloc[0]

        count_row = ablation_df[
            (
                ablation_df["Representation"]
                == "CountVectorizer"
            )
            &
            (
                ablation_df["Model"]
                == model_name
            )
        ].iloc[0]

        rq2_rows.append({

            "Model":
                model_name,

            "TF-IDF Accuracy":
                tfidf_row["Accuracy"],

            "CountVectorizer Accuracy":
                count_row["Accuracy"],

            "Accuracy Difference":
                (
                    tfidf_row["Accuracy"]
                    -
                    count_row["Accuracy"]
                ),

            "TF-IDF Weighted F1":
                tfidf_row["Weighted F1"],

            "CountVectorizer Weighted F1":
                count_row["Weighted F1"],

            "Weighted F1 Difference":
                (
                    tfidf_row["Weighted F1"]
                    -
                    count_row["Weighted F1"]
                )
        })


    rq2_comparison = pd.DataFrame(
        rq2_rows
    )

    print(
        "\nRQ2 representation differences:"
    )

    print(
        rq2_comparison.round(4)
    )

    save_dataframe(
        rq2_comparison,
        "20_rq2_representation_difference.csv"
    )


else:

    print(
        "\nRQ2 ablation disabled."
    )


# ============================================================
# EXPERIMENT CONFIGURATION
# ============================================================

print_section(
    "STEP 18 — SAVING EXPERIMENT CONFIGURATION"
)

experiment_config = {

    "dataset": {
        "train_path": TRAIN_PATH,
        "test_path": TEST_PATH
    },

    "random_state":
        RANDOM_STATE,

    "label_mapping":
        LABEL_MAP,

    "preprocessing": {
        "lowercase": True,
        "punctuation_removed": True,
        "tokenisation": True,
        "english_stopword_removal": True,
        "alphabetic_token_filter": True,
        "porter_stemming": True
    },

    "tfidf": TFIDF_CONFIG,

    "count_vectorizer": COUNT_CONFIG,

    "logistic_regression": LR_CONFIG,

    "multinomial_naive_bayes": NB_CONFIG,

    "random_forest": RF_CONFIG,

    "rq2_ablation":
        RUN_RQ2_ABLATION,

    "data_handling": {
        "remove_missing_labels": True,
        "remove_within_train_duplicates": True,
        "remove_within_test_duplicates": True,
        "check_train_test_exact_overlap": True,
        "remove_overlap_from_training_only": True,
        "retain_test_partition": True,
        "new_random_80_20_split": False
    },

    "final_dataset_sizes": {
        "training": len(train),
        "testing": len(test)
    }
}

save_json(
    experiment_config,
    "21_experiment_configuration.json"
)


# ============================================================
#  CREATE FINAL DISSERTATION SUMMARY
# ============================================================

print_section(
    "STEP 19 — GENERATING DISSERTATION RESULTS SUMMARY"
)

summary_path = (
    OUTPUT_DIR
    / "22_dissertation_results_summary.txt"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "FAKE NEWS DETECTION EXPERIMENT RESULTS\n"
    )

    f.write(
        "=" * 70
        + "\n\n"
    )

    # --------------------------------------------------------
    # Dataset
    # --------------------------------------------------------

    f.write("DATASET\n")
    f.write("-" * 70 + "\n")

    f.write(
        f"Final training samples: {len(train)}\n"
    )

    f.write(
        f"Final testing samples: {len(test)}\n"
    )

    f.write(
        f"Exact train/test overlaps removed from training: "
        f"{len(overlap_texts)}\n"
    )

    f.write(
        "\n"
    )

    # --------------------------------------------------------
    # Preprocessing
    # --------------------------------------------------------

    f.write(
        "PREPROCESSING\n"
    )

    f.write(
        "-" * 70
        + "\n"
    )

    f.write(
        "Lowercasing\n"
    )

    f.write(
        "Punctuation/non-alphabetic character removal\n"
    )

    f.write(
        "Tokenisation\n"
    )

    f.write(
        "English stop-word removal\n"
    )

    f.write(
        "Alphabetic-token filtering\n"
    )

    f.write(
        "Porter stemming\n\n"
    )

    # --------------------------------------------------------
    # TF-IDF
    # --------------------------------------------------------

    f.write(
        "TF-IDF CONFIGURATION\n"
    )

    f.write(
        "-" * 70
        + "\n"
    )

    for key, value in TFIDF_CONFIG.items():

        f.write(
            f"{key}: {value}\n"
        )

    f.write("\n")

    # --------------------------------------------------------
    # Model results
    # --------------------------------------------------------

    f.write(
        "MODEL RESULTS\n"
    )

    f.write(
        "-" * 70
        + "\n"
    )

    for _, row in results_df.iterrows():

        f.write(
            f"\n{row['Model']}\n"
        )

        f.write(
            f"Accuracy: "
            f"{row['Accuracy']:.4f}\n"
        )

        f.write(
            f"Weighted Precision: "
            f"{row['Weighted Precision']:.4f}\n"
        )

        f.write(
            f"Weighted Recall: "
            f"{row['Weighted Recall']:.4f}\n"
        )

        f.write(
            f"Weighted F1: "
            f"{row['Weighted F1']:.4f}\n"
        )

        f.write(
            f"Macro Precision: "
            f"{row['Macro Precision']:.4f}\n"
        )

        f.write(
            f"Macro Recall: "
            f"{row['Macro Recall']:.4f}\n"
        )

        f.write(
            f"Macro F1: "
            f"{row['Macro F1']:.4f}\n"
        )

    # --------------------------------------------------------
    # Best model
    # --------------------------------------------------------

    f.write(
        "\n\nBEST MODEL\n"
    )

    f.write(
        "-" * 70
        + "\n"
    )

    f.write(
        f"Model: {best_model}\n"
    )

    f.write(
        f"Accuracy: "
        f"{best_model_row['Accuracy']:.4f}\n"
    )

    f.write(
        f"Weighted F1: "
        f"{best_model_row['Weighted F1']:.4f}\n"
    )

    # --------------------------------------------------------
    # RQ answers
    # --------------------------------------------------------

    f.write(
        "\n\nRESEARCH QUESTION INTERPRETATION\n"
    )

    f.write(
        "-" * 70
        + "\n"
    )

    f.write(
        "RQ1:\n"
    )

    f.write(
        "Classical supervised machine-learning models were "
        "evaluated using a common TF-IDF representation. "
        "The resulting accuracy, precision, recall and F1 "
        "scores demonstrate their effectiveness for binary "
        "fake-news classification on the selected dataset.\n\n"
    )

    f.write(
        "RQ2:\n"
    )

    f.write(
        "TF-IDF was evaluated against CountVectorizer through "
        "a controlled representation ablation. Differences in "
        "accuracy and F1-score indicate how the weighting "
        "scheme affected classification performance.\n\n"
    )

    f.write(
        "RQ3:\n"
    )

    f.write(
        f"The best-performing classifier under the primary "
        f"accuracy criterion was {best_model}. "
        f"Its independent test-set accuracy was "
        f"{best_model_row['Accuracy']:.4f}.\n"
    )

print(
    f"Dissertation summary saved to:\n{summary_path}"
)


# ============================================================
#  FINAL RESULTS DISPLAY
# ============================================================

print_section(
    "FINAL EXPERIMENT RESULTS"
)

display_columns = [
    "Model",
    "Accuracy",
    "Weighted Precision",
    "Weighted Recall",
    "Weighted F1",
    "Macro Precision",
    "Macro Recall",
    "Macro F1"
]

final_display = results_df[
    display_columns
].copy()

final_display[
    display_columns[1:]
] = (
    final_display[
        display_columns[1:]
    ]
    * 100
)

print(
    final_display.round(2).to_string(
        index=False
    )
)


# ============================================================
# OUTPUT FILE INVENTORY
# ============================================================

print_section(
    "OUTPUT FILES"
)

for file in sorted(
    OUTPUT_DIR.iterdir()
):

    print(
        file.name
    )


# ============================================================
# FINAL COMPLETION MESSAGE
# ============================================================

print("\n")
print("=" * 80)
print("EXPERIMENT COMPLETED SUCCESSFULLY")
print("=" * 80)

print(
    f"\nAll outputs have been saved to:"
)

print(
    OUTPUT_DIR
)

print(
    "\nBest model:",
    best_model
)

print(
    "Best accuracy:",
    f"{best_model_row['Accuracy'] * 100:.2f}%"
)

print(
    "Best weighted F1:",
    f"{best_model_row['Weighted F1'] * 100:.2f}%"
)

print(
    "\nThe experiment is reproducible using the configuration file:"
)

print(
    OUTPUT_DIR
    / "21_experiment_configuration.json"
)

print(
    "\nKey research-question outputs:"
)

print(
    "RQ1 → 11_model_comparison_metrics.csv"
)

print(
    "RQ2 → 17_rq2_tfidf_ablation.csv"
)

print(
    "RQ3 → 11_model_comparison_metrics.csv"
)

print(
    "\nDone."
)
```
